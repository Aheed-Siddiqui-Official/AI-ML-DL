```python

# Select categorical columns
categorical_cols = ["CentralAir"]

# Initialize encoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

# Fit and transform
encoded_array = encoder.fit_transform(train_data[categorical_cols])

# Create DataFrame of encoded values
encoded_df = pd.DataFrame(
    encoded_array,
    columns=encoder.get_feature_names_out(categorical_cols)
)

# Reset index to avoid alignment issues
encoded_df.index = train_data.index

# Drop original categorical columns
train_data = train_data.drop(columns=categorical_cols)

# Concatenate encoded columns
train_data = pd.concat([train_data, encoded_df], axis=1)
train_data.head()
```

## Part A: Decision Tree Algorithm

We will train a Decision Tree Classifier on the house price data using four different hyperparameter combinations:

| # | Criterion | max_depth |
|---|-----------|----------|
| 1 | entropy   | 4        |
| 2 | entropy   | 5        |
| 3 | gini      | 4        |
| 4 | gini      | 5        |

Since `SalePrice` is a continuous variable and entropy/gini are classification criteria, we first convert `SalePrice` into price categories (Low, Medium, High) based on percentile ranges. This way we can properly use a classifier here.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd
import numpy as np
```

```python
X_train = pd.read_csv("X_train_70.csv")
X_test = pd.read_csv("X_test_70.csv")
```

```python
X_train.shape
```

### Preparing Features and Target

```python
y_train = X_train["PriceCategory"]
y_test = X_test["PriceCategory"]
x_train  =X_train.drop(["SalePrice","PriceCategory"], axis = 1)
x_test  =X_test.drop(["SalePrice","PriceCategory"], axis = 1)
```

```python
y_test.unique()
```

```python
from joblib import dump
```

### Model 1 — Criterion: Entropy, max_depth: 4

```python
def DecTree(x_train, y_train, x_test, y_test, criterion, max_depth,split_size):
  dt1 = DecisionTreeClassifier(criterion=criterion, max_depth=max_depth, random_state=42)
  dt1.fit(x_train, y_train)
  y_pred1 = dt1.predict(x_test)
  filename = f"DT_{criterion}_{max_depth}_{split_size}.joblib"
  dump(dt1,filename)
  return(f'Accuracy: {accuracy_score(y_test, y_pred1):.4f}')
dectree70 = DecTree(x_train, y_train, x_test, y_test,"entropy",4,70)
print(dectree70)
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
   "criterion" : ["gini","entropy"],
   "max_depth" : [4,5]
}
dt = DecisionTreeClassifier(random_state = 42)
grid_search = GridSearchCV(dt,param_grid,cv=10, scoring = "accuracy")

grid_search.fit(x_train,y_train)
```

```python
print("All combinations tried:")
cv_results = pd.DataFrame(grid_search.cv_results_)[['param_criterion', 'param_max_depth', 'mean_test_score']]
cv_results.columns = ['Criterion', 'max_depth', 'Avg CV Accuracy']
print(cv_results.to_string(index=False))

print(f"\nBest Parameters: {grid_search.best_params_}")
print(f"Best CV Accuracy: {grid_search.best_score_:.4f}")
print(f"Test Accuracy: {grid_search.score(x_test, y_test):.4f}")
```
