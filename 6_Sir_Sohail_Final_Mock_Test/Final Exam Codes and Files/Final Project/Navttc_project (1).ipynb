## Part A: Decision Tree Algorithm

We will train a Decision Tree Classifier on the house price data using four different hyperparameter combinations:

| # | Criterion | max_depth |
|---|-----------|----------|
| 1 | entropy   | 4        |
| 2 | entropy   | 5        |
| 3 | gini      | 4        |
| 4 | gini      | 5        |

Since `SalePrice` is a continuous variable and entropy/gini are classification criteria, we first convert `SalePrice` into price categories (Low, Medium, High) based on percentile ranges. This way we can properly use a classifier here.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd
import numpy as np
```

```python
def train_test(split_size):
  X_train_path = f"/content/X_train_{split_size}.csv"
  X_test_path = f"/content/X_test_{split_size}.csv"
  X_train = pd.read_csv(X_train_path)
  X_test = pd.read_csv(X_test_path)
  return X_train,X_test
# for split_size in [70,75,80]:
#   X_train,X_test = train_test(split_size)

# X_train70,X_test70 = train_test(70)
# X_train75,X_test75 = train_test(75)
# X_train80,X_test80 = train_test(80)

# print(X_train70.shape)
# print(X_test70.shape)
# print(X_train75.shape)
# print(X_test75.shape)
# print(X_train80.shape)
# print(X_test80.shape)
```

```python
def feature_target(X_train,X_test):
  y_train = X_train["PriceCategory"]
  y_test = X_test["PriceCategory"]
  x_train  =X_train.drop(["SalePrice","PriceCategory"], axis = 1)
  x_test  =X_test.drop(["SalePrice","PriceCategory"], axis = 1)
  return x_train,y_train,x_test,y_test
```

### Preparing Features and Target

```python
from joblib import dump
```

### Model 1 — Criterion: Entropy, max_depth: 4

```python
def DecTree(x_train, y_train, x_test, y_test, criterion, max_depth,split_size):
  dt1 = DecisionTreeClassifier(criterion=criterion, max_depth=max_depth, random_state=42)
  dt1.fit(x_train, y_train)
  y_pred1 = dt1.predict(x_test)
  filename = f"/content/DT_{criterion}_{max_depth}_{split_size}.joblib"
  dump(dt1,filename)
  return(f'Accuracy: {accuracy_score(y_test, y_pred1):.4f}')
```

```python
split = [70,75,80]
max_depth =[4,5]
criterion_options = ["entropy","gini"]
for split_size in split:
    # STEP 1: load dataset
    X_train, X_test = train_test(split_size)

    # STEP 2: split features and target
    x_train, y_train, x_test, y_test = feature_target(X_train, X_test)

    for depth in max_depth:
      for criterion_name in criterion_options: # Use the new variable name
      # STEP 3: train + evaluate model
        result = DecTree(
            x_train,
            y_train,
            x_test,
            y_test,
            criterion_name,
            depth,
            split_size
        )

        print(f" at {depth} max_depth, {criterion_name} criterion, {split_size}% Split -> {result}")
```

```python
from sklearn.svm import SVC
```

```python
def svm(x_train, y_train, x_test, y_test,kernel,split_size):
  svm = SVC(kernel=kernel, random_state=42)
  svm.fit(x_train, y_train)
  y_pred1 = svm.predict(x_test)
  filename = f"/content/Svm_{kernel}_{split_size}.joblib"
  dump(svm,filename)
  return(f'Accuracy: {accuracy_score(y_test, y_pred1):.4f}')
```

```python
split = [70,75,80]
kernel_names = ["linear","poly","rbf"]
for split_size in split:
    # STEP 1: load dataset
    X_train, X_test = train_test(split_size)

    # STEP 2: split features and target
    x_train, y_train, x_test, y_test = feature_target(X_train, X_test)

    for kernel_name in kernel_names: # Use the new variable name
      # STEP 3: train + evaluate model
        result = svm(
            x_train,
            y_train,
            x_test,
            y_test,
            kernel_name,
            split_size
        )

        print(f" {kernel_name} Svm {split_size}% Split -> {result}")
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {
    "criterion" : ["gini","entropy"],
    "max_depth" : [4,5]
}
dt = DecisionTreeClassifier(random_state = 42)
grid_search = GridSearchCV(dt,param_grid,cv=10, scoring = "accuracy")

grid_search.fit(x_train,y_train)
```

```python
print("All combinations tried:")
cv_results = pd.DataFrame(grid_search.cv_results_)[['param_criterion', 'param_max_depth', 'mean_test_score']]
cv_results.columns = ['Criterion', 'max_depth', 'Avg CV Accuracy']
print(cv_results.to_string(index=False))

print(f"\nBest Parameters: {grid_search.best_params_}")
print(f"Best CV Accuracy: {grid_search.best_score_:.4f}")
print(f"Test Accuracy: {grid_search.score(x_test, y_test):.4f}")
```
