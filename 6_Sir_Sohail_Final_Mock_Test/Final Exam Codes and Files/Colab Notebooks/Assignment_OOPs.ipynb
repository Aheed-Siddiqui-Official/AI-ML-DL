```python
class Device:
  __device_type = "Device"

# Attributes:

  def __init__(self, brand, model, price, color):
    self.__brand = brand # Private Attributes
    self.__model = model # Private Attributes
    self.__price = price # Private Attributes
    self.__color = color # Private Attributes

# Methods:

  def get_brand(self):
    return self.__brand

  def get_model(self):
    return self.__model

  def get_price(self):
    return self.__price

  def get_color(self):
    return self.__color

  def set_device_type(self, devType):
    self.__device_type = devType

  def get_device_type(self):
    return self.__device_type

  def get_info(self):
    print(f"Brand: {self.get_brand()}, Model: {self.get_model()}, Price: {self.get_price()}, Color: {self.get_color()}")

  def poweron(self, txt="The"):
    print(f"{txt} {self.get_device_type()} is powering on.")

  def poweroff(self, txt="The"):
    print(f"{txt} {self.get_device_type()} is powering off.")

  def charging(self, txt="The"):
    print(f"{txt} {self.get_device_type()} is charging.")
```

```python
class Mobile(Device):

  def __init__(self, brand, model, price, color):
    super().__init__(brand, model, price, color)
    self.set_device_type("Mobile")


  def camera(self, pixels=8):
    print(f"The {pixels} mega pixels camera is power on.")


class Laptop(Device):

  def __init__(self, brand, model, price, color):
    super().__init__(brand, model, price, color)
    self.set_device_type("Laptop")


  def Touchpad(self):
    print("Touchpad is power on.")
```

# Object 1

```python
mymobile = Mobile("IPhone", "IPhone-17", 200000, "Black")
```

```python
mymobile.get_brand()
```
