```python
from nltk.tokenize import word_tokenize
```

```python
sample_text = "This is a text ready to tokenize"
token = word_tokenize(sample_text)
print(sample_text)
print(token)
```

```python
from nltk.tokenize import TweetTokenizer
sample_text = "This is a tweet @jack #NLP"
tweet_tokenizer = TweetTokenizer()
token = tweet_tokenizer.tokenize(sample_text)
print(sample_text)
print(token)
```

```python
from nltk.corpus import stopwords
nltk.download("stopwords")
stopswords_ = set(stopwords.words("english"))
```

```python
sample_text = "this is sample text"
tokens = sample_text.split()
clean_tokens = [t for t in tokens if not t in stopswords_]
clean_text = " ".join(clean_tokens)
print(clean_tokens)
print(clean_text)
```

## Porter Stem

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = ["running", "flies", "happily", "better", "jumps", "happiest"]
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

## Lancaster Stem

```python
from nltk.stem import LancasterStemmer
stemmer = LancasterStemmer()
```

```python
words = ["running", "flies", "happily", "better", "jumps", "happiest"]
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

## Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
stemmer = SnowballStemmer("english")
```

```python
words = ["running", "flies", "happily", "better", "jumps", "happiest"]
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

## RegExp Stemmer

```python
from nltk.stem import RegexpStemmer
stemmer = RegexpStemmer("ing$|es$|ly$|er$", min=4)
```

```python
stemmed_words = [stemmer.stem(word) for word in words]
print(stemmed_words)
```

## WordNet Lamitizer

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
print(lemmatized_words)
```

```python
sentence = "The quick brown fox jump over the lazy dog"
words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)
print("")
print("POS Tags\n", pos_tags, "\n")

for word, tag in pos_tags:
    print(nltk.help.upenn_tagset(tagpattern=tag))
```

```python
sentence = "The big brown dog barked loudly."
words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)

chunked = nltk.ne_chunk(pos_tags)

reg_exp = "NP: {<DT>?<JJ>*<NN>*}"
rp = nltk.RegexpParser(reg_exp)
result = rp.parse(chunked)

result
```
