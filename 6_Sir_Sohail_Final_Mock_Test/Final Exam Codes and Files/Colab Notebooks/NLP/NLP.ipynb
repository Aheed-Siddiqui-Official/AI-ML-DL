## Punctuation & Special Character

```python
text = "Hello, world! This is?* 💜an&|~^+%\" example- of text preprocessing"
text
```

```python
punctuation_pattern = r'[^\w\s]'
text_cleaned = re.sub(punctuation_pattern, '', text)
print(text_cleaned)
```

```python
from nltk.corpus import stopwords
```

```python
stopwords.words("english")
```

```python
def remove_stopwords(text, language):
  stop_words = set(stopwords.words(language))
  text = text.lower()
  words_token = text.split()
  filtered_text = [word for word in words_token if word.lower() not in stop_words]
  print(language)
  print(filtered_text)
```

```python
en_text = "This is a sample sentence and we are going to remove the stopwords."
remove_stopwords(en_text, "english")
```

## URLS

```python
def remove_urls(text):
  url_pattern = re.compile(r"https?://\S+|www\.\S+")
  return url_pattern.sub(r"", text)
```

```python
text = "I hope it will be a useful article for you. Follow me: https://abc.com/@def"
remove_urls(text)
```

## HTML Tags

```python
text = """<html><div>
<h1> Natural Language Processing</h1>
<p> Text Preprocessing for NLP</p>
<a href="htps://abc.com/@def"> ABC account </a>
</div></html>"""

html_tags_pattern = re.compile(r"<.*?>")

text_without_html_tags = re.sub(html_tags_pattern, "", text)
print(text_without_html_tags)
```
