```python
def Calc(a,b):
  sm = a + b
  dif = a - b
  m = a * b
  div = a / b
  p = a ** b
  return sm, dif, m, div, p
```

```python
print(Calc(6,3))
```

```python
z = Calc(6,3)[3]
z
```

```python
s, df, m, dv, pw = Calc(6,3)
```

```python
print(s)
print(df)
print(m)
print(dv)
print(pw)
```
