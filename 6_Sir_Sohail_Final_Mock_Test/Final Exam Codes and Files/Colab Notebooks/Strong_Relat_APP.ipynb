```python
class Engine:
  def __init__(self, engnID, engn_status):
    self.__engnID = engnID
    self.__engn_status = engn_status

  def set_engine_status(self, engn_status):
      self.__engn_status = engn_status

  def get_engine_status(self):
    return self.__engn_status

  def get_engnID(self):
    return self.__engnID
```

```python
class Vehicle:
  __vehType = "Vehicle"
  def __init__(self, vehID,wheels, make, model, color, engine):
    self.__vehID = vehID
    self.__wheels = wheels    # Private Instance
    self.__make = make        # Private Instance
    self.__model = model      # Private Instance
    self.__color = color      # Private Instance
    self.__engine = engine    # Private Instance

  def set_vehType(self, vehType): # setter are usually "procedures"
    self.__vehType = vehType

  def get_vehType(self):
    return self.__vehType

  def set_vehID(self, vehID):
      self.__vehID = vehID

  def get_vehID(self):
      return self.__vehID

  def set_color(self, color):
      self.__color = color

  def get_color(self):
      return self.__color

  def get_engine(self):
      return self.__engine

  def get_wheels(self):
    return self.__wheels

  def get_info(self):
    return f"Wheels : {self.__wheels}, Make : {self.__make}, Model : {self.__model}, Color : {self.__color}"

  def move(self, txt="The"):
    print(f"{txt} {self.__vehType} is moving.")
```

```python
engine = Engine("E-1789", "Working")
```

```python
mycar = Vehicle("C-987", 4, "Toyota", "Camry", "Blue", engine)
```

```python
info = mycar.get_info()
print(info)
```

```python
mycar.set_vehType("Car")
print(f"My {mycar.get_vehType()} is being driven")
print(f"My Engine Status is: {engine.get_engine_status()}")
```
