Outliers

```python
# IQR method
def outlier_remover(df, column):
  q1 = df[column].quantile(0.25)
  q3 = df[column].quantile(0.75)
  IQR = q3-q1
  lower_bound = q1-1.5*IQR
  upper_bound = q3+1.5*IQR
  df = df[(df[column] > lower_bound) & (df[column] < upper_bound)]
  return df
```

```python
col = ["amount", "oldbalanceOrg", "newbalanceOrig"]
```

```python
for x in col:
  plt.figure(figsize=(8, 6))
  sns.boxplot(x=df[x])
  plt.title(f'boxplot of {x} Before Removing Outliers')
  plt.xlabel(x)
  plt.ylabel('Frequency')
  plt.show()
```

```python
for x in col:
  df = outlier_remover(df, x)
```

```python
for x in col:
  plt.figure(figsize=(8, 6))
  sns.boxplot(x=df[x])
  plt.title(f'boxplot of {x} After Removing Outliers')
  plt.xlabel(x)
  plt.ylabel('Frequency')
  plt.show()
```

In this dataset, I have cleaned nan values and remove outliers.
