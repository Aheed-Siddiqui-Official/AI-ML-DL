```python
class Car:
  vehType = "Car" # Class variable

  def __init__(self, wheels, make, model, color):
    self.wheels = wheels # Instance Attribute
    self.make = make # Instance Attribute
    self.model = model # Instance Attribute
    self.color = color # Instance Attribute

  def move(self, txt="The"):
    print(f"{txt} {self.color} {self.model} {self.vehType} is moving.")

```

## INHERITANCE

```python
class Vehicle:
  __vehType = "Vehicle"

  def __init__(self, wheels, make, model, color):
    self.__wheels = wheels    # Private Instance
    self.__make = make        # Private Instance
    self.__model = model      # Private Instance
    self.color = color        # Public Instance

  def set_vehType(self, vehType): # setter are usually "procedures"
    self.__vehType = vehType

  def get_vehType(self):
    return self.__vehType

  def get_wheels(self):
    return self.__wheels
  def get_info(self):
    return f"Wheels : {self.__wheels}, Make : {self.__make}, Model : {self.__model}, Color : {self.color}"
  def move(self, txt="The"):
    print(f"{txt} {self.__vehType} is moving.")
```

## Derived class

```python
class Car(Vehicle):
  def __init__(self, wheels, make, model, color):
    super().__init__(wheels, make, model, color)
    self.set_vehType("Car")

class Motorcycle(Vehicle):
  def __init__(self, wheels, make, model, color):
    super().__init__(wheels, make, model, color)
    self.set_vehType("Motorcycle")


```

## Method Overiding

```python
class Vehicle:
  __vehType = "Vehicle"

  def __init__(self, wheels, make, model, color):
    self.__wheels = wheels    # Private Instance
    self.__make = make        # Private Instance
    self.__model = model      # Private Instance
    self.color = color        # Public Instance

  def set_vehType(self, vehType): # setter are usually "procedures"
    self.__vehType = vehType

  def get_vehType(self):
    return self.__vehType

  def get_wheels(self):
    return self.__wheels
  def get_info(self):
    return f"Wheels : {self.__wheels}, Make : {self.__make}, Model : {self.__model}, Color : {self.color}"
  def move(self, txt="The"):
    print(f"{txt} {self.__vehType} is moving.")
```

```python
class Car(Vehicle):
  def __init__(self, wheels, make, model, color):
    super().__init__(wheels, make, model, color)
    self.set_vehType("Car")
  def move(self, txt="The"):
    print(f"{txt} {self.get_vehType()} is being drivien.")

class Motorcycle(Vehicle):
  def __init__(self, wheels, make, model, color):
    super().__init__(wheels, make, model, color)
    self.set_vehType("Motorcycle")
  def move(self, txt="The"):
    print(f"{txt} {self.get_vehType()} is being ridden.")
```

```python
mycar = Car(4, "Toyota", "Camry", "Blue")
mybike = Motorcycle(2, "Yamaha", "MT-07", "Red")
```

```python
mycar.move("My")
```
