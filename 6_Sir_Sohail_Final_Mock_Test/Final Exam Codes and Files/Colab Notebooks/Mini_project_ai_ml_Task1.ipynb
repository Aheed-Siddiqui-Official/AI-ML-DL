```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings('ignore')
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
train_data = pd.read_csv('house_price_classification.csv')
train_data.head()
```

```python
train_data.shape
```

```python
train_data
```

```python
train_cols =  train_data.columns
missing_train_cols = []
missing_train_vals = []
for col in train_cols:
  if train_data[col].isnull().sum() > 0:
    missing_train_cols.append(col)
    missing_train_vals.append(train_data[col].isnull().sum())
    print(f"column {col} have {train_data[col].isnull().sum()} missing values.")
```

```python
a = ["LotFrontage", "MasVnrArea", "GarageYrBlt"]
for i in a:
  train_data[i].fillna(train_data[i].median(), inplace=True)
```

```python
train_data["Electrical"].fillna(train_data["Electrical"].mode()[0], inplace=True)
```

```python
bsmt_cols = ["BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2"]
for i in bsmt_cols:
  train_data[i].fillna("None", inplace=True)
```

```python
garage_cols = ["GarageType", "GarageFinish", "GarageQual", "GarageCond"]
for i in garage_cols:
  train_data[i].fillna("None", inplace=True)
```

```python
high_missing_cols = ["Alley", "FireplaceQu", "Fence", "PoolQC", "MiscFeature"]
for col in high_missing_cols:
  train_data[col].fillna("None", inplace=True)
```

```python
train_data["MasVnrType"].fillna("None", inplace=True)
```

```python
train_data.isnull().sum().sum()
```

```python
categorical_cols = train_data.select_dtypes(include='object').columns
```

```python
categorical_cols = categorical_cols.tolist()
categorical_cols = categorical_cols.remove("SaleCondition")
```

```python
one_SaleCondition = ["SaleCondition"]
```

```python
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
encoded_data = ohe.fit_transform(train_data[one_SaleCondition])
encoded_df = pd.DataFrame(encoded_data, columns=ohe.get_feature_names_out(one_SaleCondition))
train_data = train_data.reset_index(drop=True)
train_data = pd.concat([train_data, encoded_df], axis=1)
train_data.drop(one_SaleCondition, axis=1, inplace=True)
```

```python
label_encoders = {}

categorical_cols = train_data.select_dtypes(include='object').columns.tolist()

for col in categorical_cols:
  le = LabelEncoder()
  train_data[col] = le.fit_transform(train_data[col].astype("str"))
  label_encoders[col] = le
```

```python
train_data.to_csv("/content/house_price_classification_cleaned.csv", index=False)
```
