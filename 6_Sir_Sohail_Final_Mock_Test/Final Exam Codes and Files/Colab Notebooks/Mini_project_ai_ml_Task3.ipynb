```python
import pandas as pd
import numpy as mp
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from joblib import dump
```

```python
df_train_70 = pd.read_csv('clean_train_70.csv')
df_train_75 = pd.read_csv('clean_train_75.csv')
df_train_80 = pd.read_csv('clean_train_80.csv')
df_test_20 = pd.read_csv('clean_test_20.csv')
df_test_25 = pd.read_csv('clean_test_25.csv')
df_test_30 = pd.read_csv('clean_test_30.csv')
```

```python
def task_3(X_train, y_train, X_test, y_test, criteria, tree_depth, split_size):
  dt = DecisionTreeClassifier(criterion=criteria, max_depth=tree_depth, random_state=42)
  dt.fit(X_train, y_train)
  y_pred = dt.predict(X_test)
  accuracy = accuracy_score(y_test, y_pred)
  print(f"Accuracy: {accuracy:.4f}")

  filename = f'/content/DT_{criteria}_{tree_depth}_{split_size}.joblib'
  dump(dt, filename)
  print(f"Model saved as: {filename}")

  return dt, accuracy
```

```python
def evaluate_all_splits(df_train_70, df_test_30,
                        df_train_75, df_test_25,
                        df_train_80, df_test_20,
                        target_column='SalePrice',
                        criteria='entropy', tree_depth=5):
  splits = [
      (df_train_70, df_test_30, '70_30'),
      (df_train_75, df_test_25, '75_25'),
      (df_train_80, df_test_20, '80_20')
  ]

  for train_df, test_df, split_label in splits:
    print(f"\n--- Evaluating model for split: {split_label} ---")

    X_train = train_df.drop(columns=[col for col in ['Unnamed: 0.1', 'Unnamed: 0', target_column] if col in train_df.columns])
    y_train = train_df[target_column]

    X_test = test_df.drop(columns=[col for col in ['Unnamed: 0.1', 'Unnamed: 0', target_column] if col in test_df.columns])
    y_test = test_df[target_column]
    model, accuracy = task_3(X_train, y_train, X_test, y_test, criteria, tree_depth, split_label)
    print(f"Model for {split_label} split achieved an accuracy of: {accuracy:.4f}")

  print("\n--- Evaluation complete for all splits ---")
```

Now, let's call the function `evaluate_all_splits` using the loaded dataframes. We'll use `SalePrice` as the target variable.

```python
from joblib import dump # Added to resolve NameError: name 'dump' is not defined
evaluate_all_splits(df_train_70, df_test_30,
                    df_train_75, df_test_25,
                    df_train_80, df_test_20,
                    target_column='SalePrice')
```
