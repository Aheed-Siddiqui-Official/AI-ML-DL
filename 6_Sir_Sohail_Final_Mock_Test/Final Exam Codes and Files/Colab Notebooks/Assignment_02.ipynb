```python
# Question No 06

a = 5
b = 6
c = 7

S = (a + b + c) / 2

area = (S*(S-a)*(S-b)*(S-c)) ** 0.5

print(f"The area of triangle is {round(area, 2)}.")
```

```python
# Question No 07

import cmath

a = 1
b = 5
c = 6

Discriminant = b**2 - 4 * a * c

sol1 = (-b-cmath.sqrt(Discriminant))/(2 * a)
sol2 = (-b+cmath.sqrt(Discriminant))/(2 * a)

print(f"The solutions are {sol1} and {sol2}.")

```
