```python
class Engine:
  def __init__(self, engID):
      self.__engID = engID
      self.__estatus = "Working"

  def set_estatus(self, estatus):
      self.__estatus = estatus

  def get_engID(self):
      return self.__engID

  def get_estatus(self):
      return self.__estatus
```

```python
class Vehicle:
  def __init__(self, vehID, wheels, make, model, color, engine):
      self.__vehType = "vehType"
      self.__vehID = vehID
      self.__wheels = wheels
      self.__make = make
      self.__model = model
      self.__color = color
      self.__engine = engine


  def set_vehType(self, vehType):
      self.__vehType = vehType
  def set_color(self, color):
      self.__color = color
  def set_engn_status(self, sts="Working"):
      self.__engine.set_estatus(sts)


  def get_vehType(self): return self.__vehType
  def get_eng(self): return self.__engine

  def get_engn_status(self): return self.__engine.get_estatus()
  def get_engID(self): return self.__engine.get_engID()

  def get_info(self):
      info = (f" {self.__make} {self.__model} {self.__color} {self.__wheels} wheeler")
      return info

  def move(self, txt="The"):
      print(f"{txt} {self.get_vehType()} is being driven")
```

```python
class Car(Vehicle):
  def __init__(self, vehID, wheels, make, model, color, engine):
    super().__init__(vehID, "Car", wheels, make, model, color, engine)
  def move(self, txt="The"):
    engn = super().get_eng()
    current_status = engn.get_estatus()
    if current_status != "Working":
      raise Exception(f"The {self.get_vehType()} couldn't start: Engine {current_status}")
    else:
      print(f"The {txt} {self.get_vehType()} is being driven")
```
