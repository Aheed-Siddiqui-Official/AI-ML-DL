# E-Commerce System

```python
class Product:
  __Product = "Product"
  def __init__(self, name, price, description):
      self.__name = name
      self.__price = price
      self.__description = description

  def set_product_name(self, name):
      self.__name = name

  def get_product_name(self):
      return self.__name

  def get_name(self):
      return self.__name
  def get_price(self):
      return self.__price
  def get_description(self):
      return self.__description
```

```python
class Physical_Product(Product):
  def __init__(self, name, price, description, weight, shipping_cost):
    super().__init__(name, price, description)
    self.weight = weight
    self.shipping_cost = shipping_cost
  def get_cost(self):
    return self.get_price() + self.shipping_cost
```

```python
class Digital_Product(Product):
  def __init__(self, name, price, description, file_size, download_link):
    super().__init__(name, price, description)
    self.file_size = file_size
    self.download_link = download_link

  def get_cost(self):
    return self.get_price()
```

```python
class ShoppingCart:
  def __init__(self):
    self.items = []

  def add_product(self, product):
    self.items.append(product)

  def calculate_total(self):
    total = 0
    for item in self.items:
        total += item.get_cost()
    return total
```

```python
p1 = Physical_Product("Laptop", 1000, "Gaming laptop", 2.5, 50)
p2 = Digital_Product("Ebook", 20, "Python book", "5MB", "link.com")
```

```python
cart = ShoppingCart()
cart.add_product(p1)
cart.add_product(p2)

# Total cost
print("Total Cost:", cart.calculate_total())
```
