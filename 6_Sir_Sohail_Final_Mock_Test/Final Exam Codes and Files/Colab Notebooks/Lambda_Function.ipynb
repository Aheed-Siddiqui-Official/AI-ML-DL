## Lamda (function name = lambda Input : Output)

```python
Add10 = lambda a : a + 10
```

```python
x = Add10(5)
x
```

```python
multiply = lambda a, b : a * b
```

```python
multiply(4, 5)
```

```python
even_fn = lambda x : True if x%2 == 0 else False
```

```python
even_fn(4)
```

```python
List = [1,3,2,5,20]
EvnLSt = []
OddLst = []
for i in List:
  if even_fn(i) == True:
    EvnLSt.append(i)
  else:
    OddLst.append(i)
```

```python
print(EvnLSt)
print(OddLst)
```

```python
r_Even = list(filter(even_fn, List))
```

```python
print(r_Even)
```

```python
r_Even = list(filter(lambda x : x %2 == 0, List))
```

```python
r_Even
```

```python
r_Even = list(map(lambda i : i **2, List))
r_Even
```

Lambda Funtion Within Method

```python
def myfunc():
  return lambda a, b : a * b
```

```python
multiply = myfunc()
```

```python
multiply(2, 3)
```
