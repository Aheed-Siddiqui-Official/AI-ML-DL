```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('smartphone_sales.csv')
```

```python
df01 = df.iloc[:, 1:]
```

```python
df01
```

```python
df02 = df01.sum()
df02
```

# Bar Plot

```python
df02.values
```

```python
plt.figure(figsize=(15,6))
plt.xticks(range(len(df02)), df01.columns)
plt.title("Bar plot")
plt.xlabel("Phone")
plt.ylabel("Sales")
chrt = sns.barplot(x=df02.index, y=df02.values)

for p in chrt.patches:
  chrt.annotate(format(p.get_height(), ",.0f"),
                xy=(p.get_x() + p.get_width() / 2., p.get_height()),
                ha="center", va = "center", xytext=(0,9),
                textcoords="offset points")
```

# Line Graph

```python
plt.figure(figsize=(10,4))
plt.title("Line Graph")
plt.xlabel("Phone")
plt.ylabel("Sales")
for a,b in zip(df02.index, df02.values):
  plt.text(a,b, str(b))
sns.lineplot(data=df02, marker="o", c="blue")
```

# Pie Chart

```python
plt.figure(figsize=(10,10))
plt.title("Pie Chart")
plt.pie(df02.values, labels=df02.index, labeldistance=1.15, autopct="%1.2f%%", wedgeprops={"linewidth":3, "edgecolor":"white"})
```

# Distribution Plot

```python
sns.displot(x=df02.index, y=df02.values)
```

# Stack Plot

```python
ax = df.plot.bar(stacked="True", figsize=(15,7), rot=8)

plt.title("Stack Plot")
plt.xlabel("Phone")
plt.ylabel("Sales")
plt.xticks(range(len(df)), df["Quarter"])

for c in ax.containers:
  ax.bar_label(c, label_type="center")
ax.legend(title="Categories", bbox_to_anchor=(1.05,1), loc="upper left")
```

# Unstacked Bar Chart

```python
chrt = df01.plot(kind="bar", figsize=(15,7))
plt.title("Unstacked Bar Chart")
plt.xlabel("Phone")
plt.ylabel("Sales")
plt.xticks(range(len(df)), df["Quarter"])

for p in chrt.patches:
  chrt.annotate(format(p.get_height(), ",.0f"),
                xy=(p.get_x() + p.get_width() / 2., p.get_height()),
                ha="center", va = "center", xytext=(0,1),
                textcoords="offset points")
```
