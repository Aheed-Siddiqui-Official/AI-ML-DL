## Palin Droom

```python
txt ='python'
print(txt[::-1])
```

```python
txt =['python','civic','mongodb', 'radar']
```

```python
for i in txt:
  print(i)
```

```python
for i in txt:
  if i == i[::-1]:
    print(i)
```

```python
for i in txt:
  if i == i[::-1]:
    res = "\b"
  else:
    res = "NOT"
  print(f"{i} is {res} a Palindrome")
```

```python
txt =['python','civic','mongodb', 'radar']
res=[]
for i in txt:
  if i == i[::-1]:
    res.append(i)
print(res)
```

```python
x = [2, 4, 6, 7, 8]
fnd = int(input("Enter a Number: "))
for i in x:
  if i == fnd:
    output = "Found"
    break
  else:
    output = "Not Found"
print(fnd, output)
```

```python
print("i j")
for i in range(1,6,1):
  for j in range(2,4,1):
    print(i,j)
  print("-----")
```

```python
for m in range(1,6,1):
  for t in range(2,4,1):
    print(t, "x", m, "=", t*m, end = "\t")
  print("")

```
