# Method Resolution Order

```python
class A:
  def process(self):
    print("A process()")
```

```python
class B:
  pass
```

```python
class C(A, B):
  pass
```

```python
obj = C()
```

```python
C.mro()
```

```python
obj.process()
```

```python
class B:
  def process(self):
    print("B process()")
```

```python
class C(A, B):
  pass
```

```python
obj = C()
obj.process()
C.mro()
```

```python
class C(B, A):
  pass
```

```python
obj = C()
obj.process()
C.mro()
```

```python
class C(A, B):
  def process(self):
    print("C process()")
```

```python
class D(C,B):
  pass
```

```python
obj = D()
obj.process()
D.mro()
```

```python
class A:
  def process(self):
    print("A process()")
class B(A):
  pass
class C(A):
  def process(self):
    print("C process()")
class D(B,C):
  pass
```

```python
obj = D()
obj.process()
D.mro()
```

```python
class A:
  def process(self):
    print("A process()")
class B(A):
  def process(self):
    print("B process()")
class C(A, B): # Here A and B are same thing : Error
  pass
```
