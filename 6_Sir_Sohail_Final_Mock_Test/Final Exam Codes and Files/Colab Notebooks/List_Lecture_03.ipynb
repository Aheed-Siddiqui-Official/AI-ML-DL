## Mathematical Functions

```python
ages = [23, 16, 14, 28, 19, 11, 38]
```

```python
youngest = min(ages)
eldest = max(ages)
total_ages = sum(ages)
count = len(ages)
average = total_ages / count

print(f"Youngest age is {youngest}.")
print(f"Eldest age is {eldest}.")
print(f"Total ages is {total_ages}.")
print(f"Count is {count}.")
print(f"The average age is {round(average, 1)}.")
```

```python
ages.sort(reverse=True)
```
