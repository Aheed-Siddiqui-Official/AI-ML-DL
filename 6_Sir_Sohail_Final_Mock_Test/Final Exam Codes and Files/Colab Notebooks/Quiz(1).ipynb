```python
nums = [1, 2, 3, 4]
squares = []

for n in nums:
  if n % 2 != 0:
    squares.append(n ** 2)

print(squares)
```

```python
class Animal:
  def __init__(self, name):
    self.name = name

  def speak(self):
    return "Animal sound"
class Dog(Animal):
  def speak(self):
    return "Woof!"
```

```python
animal = Dog("Buddy")
print(animal.speak())
```

```python
class Calculator:
  count = 0

  def __init__(self):
    Calculator.count += 1
  @classmethod
  def get_count(cls):
    return cls.count
```

```python
c1 = Calculator()
c2 = Calculator()
c3 = Calculator()

print(Calculator.get_count())
```

```python
class Person:
  def __init__(self, name):
    self.__name = name
  def get_name(self):
    return self.__name
```

```python
p = Person("Alice")
print(p.__name)
```

```python
class Vector:
  def __init__(self, x, y):
    self.x = x
    self.y = y
  def __add__(self, other):
    return Vector(self.x + other.x, self.y + other.y)
  def __str__(self):
    return f"({self.x}, {self.y})"
```

```python
v1 = Vector(2, 3)
v2 = Vector(1, 4)

print(v1 + v2)
```
