```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import math
```

```python
x = np.linspace(-10, 10, 50)
```

```python
def step(x):
  return 1 if x>=0 else 0
```

```python
binary = []
for i in x:
  y = step(i)
  binary.append(y)
binary
```

```python
y = [step(i) for i in x]
plt.plot(x, y)
plt.show()
```

## Sigmoid Function

```python
def sigmoid(x):
  y = 1/(1 + math.exp(-x))
  return y
```

```python
y = [sigmoid(i) for i in x]
plt.plot(x, y)
plt.show()
```

## Tanh Function

```python
def tanh(x):
  return (2 / (1 + math.exp(-2*x)))
```

```python
y = [tanh(i) for i in x]
plt.plot(x, y)
plt.show()
```

## ReLU Function

```python
 def relu(x):
  return x if x >= 0 else 0 # max(x, 0)
```

```python
y = [relu(i) for i in x]
plt.plot(x, y)
plt.show()
```

## Leaky Relu Function

```python
def leaky_relu(x, alpha=0.5):
  return x if x >= 0 else alpha*x # max(x, alpha*x)
```

```python
y = [leaky_relu(i) for i in x]
plt.plot(x, y)
plt.show()
```

```python
binary = []
for i in x:
  y = leaky_relu(i)
  binary.append(y)
binary
```

## SoftMax Function

```python
def softmax(x): # there is an error
  e_x_sum = 0
  for i in x:
    e_x_sum += math.exp(i)
  return [math.exp(i)/e_x_sum for i in x]
```

```python
y = softmax(x)
plt.plot(x, y)
plt.show()
```

## Softplus


```python
def softplus(x):
  return np.log(1 + np.exp(x))
```

```python
y = [softplus(i) for i in x]
plt.plot(x, y)
plt.show()
```
