# Encoding

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('emp.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df2 = df.copy()
```

```python
sorted(df2["job"].unique())
```

```python
df2["job"].value_counts()
```

```python
cols_name = {"job" : {"ANALYST" : 3, "CLERK" : 5, "MANAGER" : 2, "PRESIDENT" : 1, "SALESMAN" : 4}}
```

```python
cols_name
```

```python
df2 = df2.replace(cols_name)
df2.head()
```

```python
def salCatg(salary):
  if salary >= 3000:
    r=1
  elif salary >= 1500:
    r=2
  else:
    r=3
  return r
```

```python
df2["sal"] = df2["sal"].apply(salCatg)
df2.head()
```

```python
df3 = df.copy()
df.dtypes
```

```python
df3["job"] = df3["job"].astype("category")
df3.dtypes
```

```python
sorted(df3["job"].unique())
```

```python
df3["job1"] = df3["job"].cat.codes
```

```python
df3.head()
```

```python
df3["comm"] = df3["comm"].astype("category")
```

```python
sorted(df3["comm"].unique())
```

```python
df3["comm1"] = df3["comm"].cat.codes
```
