```python
# Range Function (start, stop, step)

for i in range(0, 3, 1):
  print("Python", end = "\t")
```

```python
Number = int(input("Enter a number: "))
for i in range(1,11,1):
  print(f"{Number} x {i} = {Number * i}")
```

```python
Number = int(input("Enter a number: "))
min = 1
max = 11
for i in range(min,max,1):
  print(f"{Number} x {max - i} = {Number * (max - i)}")
```

```python
Number = int(input("Enter a number: "))
min = 0
max = 10
for i in range(max,min, -1):
  print(f"{Number} x {i} = {Number * (i)}")
```

```python
for i in range(1,11):
  if (i % 2) == 0:
    result = " Even Number"
  else:
    result = " Odd Number"
  print(f"{i}: {result}")
```

```python
for i in range(1,20):
  if  (i % 5 == 0) and (i % 3 == 0):
    print(i,"Yes and No")
  elif (i % 5 == 0) or (i % 3 == 0):
    print(i, "Yes")
  else:
    print(i, "No")
```

# Unique Code

```python
Lst = [4, 3, 6, 2, 7, 3, 5, 2]
```

```python
LstUnique = []
```

```python
for i in Lst:
  if i not in LstUnique:
    LstUnique.append(i)
print(LstUnique)
```

## Pyramid

```python
n = 6
for i in range(1, n, 1):
  print((n-i) * " " + i * "#")
  print(i * "#")
```

```python
n = 6
for i in range(1, n, 1):
  print((n-i) * " " + i * "#", end="")
  print(i * "#")
```

```python
n = 6
for i in range(1, n, 1):
  print((n-i) * " " + i * "#", end="\b")
  print(i * "#")
```

```python
n = 6
for i in range(n, 0, -1):
  print((n-i) * " " + i * "#", end="\b")
  print(i * "#")
```

```python
n = 6
for i in range(1, n, 1):
  print((n-i) * " " + i * "#", end="\b")
  print(i * "#")
for i in range(n, 0, -1):
  print((n-i) * " " + i * "#", end="\b")
  print(i * "#")
```
