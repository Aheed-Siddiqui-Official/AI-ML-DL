# Try and Except
It is used when we know there may be error with some values:



```python
def div(num1, num2):
  try:
    divide = float(num1) / float(num2)
  except:
    divide = "Can't divide by zero"
  return divide
```

```python
n1 = input("Enter a number: ")
n2 = input("Enter another number:")
c = div(n1, n2)
c
```

```python
def div(num1, num2):
  try:
    divide = float(num1) / float(num2)
  except:
    divide = "Can't divide by zero"
  finally:
    return divide
```

```python
n1 = input("Enter a number: ")
n2 = input("Enter another number:")
c = div(n1, n2)
c
```
