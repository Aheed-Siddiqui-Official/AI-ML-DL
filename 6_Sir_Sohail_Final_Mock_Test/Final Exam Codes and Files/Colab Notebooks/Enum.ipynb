# Enum
* Enums is Python are used to define a set of named constant values.
* Enumeration are created using classes.
* Enums have names and values associated with them.

```python
import enum
```

```python
class Day(enum.Enum):
  Sunday = 1
  Monday = 2
  Tuesday = 3
```

## Properties

```python
print("The string representation of enum member is : ", end="")
print(repr(Day.Sunday))
```

```python
print("The type of enum member is : ", end ="")
print(type(Day.Sunday))
```

```python
print("The name of enum member is : ", end ="")
print(Day.Sunday.name)
```

```python
print("All the enum values are: ")
for Dy in Day:
  print(Dy)
```

```python
di = {}
di[Day.Sunday] = "Holiday"
di[Day.Tuesday] = "Working day"

if di == {Day.Sunday : "Holiday", Day.Tuesday : "Working day"}:
  print("Enum is hashed")
else:
  print("Enum is not hashed")
```

```python
for Dy in di:
  print(f"{Dy.name} is {di[Dy]}")
```

```python
print("The enum member associated value 2 is : ", end ="")
print(Day(2))
```

```python
print("The enum member associated with name Tuesday is : ", end="")
print(Day["Tuesday"])
```

```python
mem = Day.Sunday
print("The value associated with Sunday is : " , end="")
print(mem.value)
```

```python
print("The value associated with Sunday is : " , end="")
print(mem.name)
```

```python
if Day.Sunday is Day.Monday:
  print("Sunday and Monday are same")
else:
  print("Sunday and Monday are different")
```

```python
if Day.Sunday != Day.Monday:
  print("Sunday and Monday are different")
else:
  print("Sunday and Monday are same")
```

```python
if Day.Sunday is not Day.Monday:
  print("Sunday and Monday are different")
else:
  print("Sunday and Monday are same")
```
