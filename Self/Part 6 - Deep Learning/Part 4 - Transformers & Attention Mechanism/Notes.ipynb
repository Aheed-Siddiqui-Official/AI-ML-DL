### Query

A Query represents what a word is trying to find in the rest of the sentence. You can think of it as a question the word is asking, such as “Which other words are important for my meaning?” When the model processes a token, it generates a Query vector that gets compared with the Key vectors of all other tokens.The stronger the match, the more attention that word receives. In simple terms, the Query is the search signal that helps a word gather the right context before forming a deeper understanding

### Key

A Key is like a label that describes what information a word contains and how relevant it might be to others. Every token produces a Key vector, and these keys are checked against incoming Queries to measure compatibility. If a Key closely aligns with a Query, the model interprets that word as important for the current context. Essentially, Keys help answer the question, “Do I have the information you are looking for?” This matching process is what allows the model to identify relationships between words, regardless of how far apart they appear in a sentence.

### Value

A Value holds the actual information that will be shared once a strong Query–Key match is found. After calculating attention scores, the model does not pass along the Keys or Queries it combines the Value vectors based on how much attention each word deserves. This weighted blending creates a richer, contextual representation of the token. Put simply, if the Query is the question and the Key determines relevance, the Value is the knowledge that gets delivered, enabling the model to understand language in a more connected and meaningful way.

## Moving to Feed Forward Neural Network

The Feed Forward Neural Network (FFN) is the second main part of an encoder layer that comes after
self-attention. Once self-attention helps each word understand its relationship with other words,the FFN processes each word individually to extract deeper patterns and meaning. It is simply a two-layer neural network (Linear → Activation → Linear) that first expands the embedding size (for example, 512 to 2048) to give the model more capacity to learn complex features, and then compresses it back to the original size. Unlike attention, there is no communication between words here the same network is applied separately to every token. In short, self-attention builds context, and the feed forward network strengthens that understanding by transforming it into a richer representation.

Multiple encoder layers are used to gradually deepen the model’s understanding of the sentence. The first layer typically captures basic relationships between words, such as nearby context, while the next layers build on that information to detect more complex patterns like long-range dependencies and semantic meaning. Each layer refines the representation further, making it more contextual and abstract. Just like deep neural networks become more powerful with depth, stacking encoder layers allows the transformer to move from simple word-level understanding to a much richer comprehension of the entire sentence.
