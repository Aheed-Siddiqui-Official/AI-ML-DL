## Theory Behind CNN / Architecture

In CNN the algorithim looks small patches and identifies the image, while ANN saw each and every pixel to identify

We have an input layer first of all

1) Convolutional Layer Detective Magnifying Glass

A small magnifying layer, small filter (3*3 window) across the picture

The filter might highlight edges, colors, or textures

e.g. one filter learns to find "straight lines", another learns "round shapes"

2) Pooling Layer The Shortcut

Once the detective finds details, we dont need to keep every tiny pixels

Pooling just keeps the most important information

Think it like zooming out a little you lose small details, but still know big shape

3) More Convolution + Pooling Layer

To build image understanding

As network goes deeper 

Middle layers finds shapes(eyes, nose, wheels)

Deeper layers find objects(a cat, car, castle)

4) Fully Connected Layer - The Decision Maker

At the end the CNN flattens everything it learned and connects it like a normal ANN

e.g: 70% chance its a cat

20% chance its a dog

10% chance its a rabbit

Fully connected layers are like ANN
