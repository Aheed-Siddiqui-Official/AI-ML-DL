#### 2) Bag of Words

Technique to convert text into numbers.

Works sentence wise not word wise

It matters only words frequency

max features means which word is repeated most time to keep it

Order doesnt matter in bag of words

#### Pros

1) Easy to implement
2) Fixed size
3) Now we can use in the ML model vector and document size remain fixed
4) Work with ML models
5) Fast and efficient

#### Cons

1) Sparse matrix lots of zeros remains
2) Out of vocabulary
3) Sementic meaning (Better than OHE)
4) Out of order
