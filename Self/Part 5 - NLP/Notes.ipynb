Approaches of NLP : 

1) Rule Based Approach (Old School Approach)

A sentence contains not good mark it as negative sentence

Not scalable for large data

2) Statistical ML Approach

Language is trated like data, and algotithms learn patterns from it

Naive Bayers

Logistic Regression

SVM

Requires text to converted into numbers using techniques like bag of words

Used for sentiment analysis, spam etc

Loses context of words like sarcasm 

3) Deep Learning Based Approach

Used from neural networks to automatically learn complex patterns and context from text.

RNN

LSTM

GRU

Transformers(BERT, GPT, T5 etc)

Advance embedding techniques

## Pipeline

### Step 1 : Gathering the text or say getting the data

1) Prebuild data set 
2) Web scrapping
3) APIS Structures
4) Manual Collection
5) Use NLP tools to AutoGenerate Data (Test Augmentation)

### Step 2 : Text Cleaning

1) Lowercasing
2) Remove punctuation they dont add context in ML context
3) Remove numbers
4) Remove URLS/Links
5) Remove HTML Tags like Web Scrapping
6) Remove emogis and Special Characters
7) Remove stopwords e.g. is, the, was, in
8) Optional (Special Characters)
