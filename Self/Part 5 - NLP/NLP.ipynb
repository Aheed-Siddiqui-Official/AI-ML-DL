# Converting output variable in numeric terms

```python
unique_emotions = df['emotion'].unique()
emotion_numbers = {}
i = 0

for emo in unique_emotions:
    emotion_numbers[emo] = i
    i+=1

df['emotion'] = df['emotion'].map(emotion_numbers)
```

# Lowercase

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

# Removing Punctuation

```python
import string

def remove_punc(txt):
    return txt.translate(str.maketrans('', '', string.punctuation))

df['text'] = df['text'].apply(remove_punc)
```

# Removing Numbers

```python
def remove_numbers(txt):
    new = ""
    for i in txt:
        if not i.isdigit():
            new = new+i
    return new

df['text'] = df['text'].apply(remove_numbers)
```

# Remove emogis and Special Characters

```python
def remove_emogis(txt):
    new = ""
    for i in txt:
        if i.isascii():
            new += i
    return new

df['text'] = df['text'].apply(remove_emogis)
```

# Remove stopwords e.g. is, the, was, in (Remove in ML context but keep in DL context)

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
# nltk.download('punkt')
# nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
  words = txt.split()
  cleaned = []
  for i in words:
    if not i in stop_words:
      cleaned.append(i)

  return ' '.join(cleaned)

df['text'] = df['text'].apply(remove)
```
