## Clusturing Algorithem

### ---> K-means clusturing

centroid based algo

different clusters on a graph seperated

1) decide n clusters
2) initialize centeroids
3) Assign clusters -> euclidian distance from center
4) Re-assign the center point
5) Now again 3rd step

For making the clusters and deciding K value you have to use Elbow Method

### Elbow method is used for defining value of K

WCSS -> within cluster sum of square distance

K = 1 one center point

d1^2+d2^2+d3^2+d4^2+d5^2 - WCSS

K = 2

WCSS1 + WCSS2 two center point

K = 3

WCSS1 + WCSS2 + WCSS3 three center point

A slanting curve will be formed in graph when assigning values called elbow method

best cluster point is when there is no change in graph at a declining point

### Disadvantages

we have to initialize the K-value

outliers will move the center point

non circular data
